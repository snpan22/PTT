```python
import os, pickle, numpy as np

root = '/storage/project/r-gchou3-0/shared/'
dbinfo_path = os.path.join(root, "waymo_processed_data_v0_5_0_waymo_dbinfos_train_sampled_1.pkl")

with open(dbinfo_path, "rb") as f:
    db = pickle.load(f)

entry = db["Vehicle"][0]
bin_path = os.path.join(root, entry["path"])

n = entry["num_points_in_gt"]
size = os.path.getsize(bin_path)

print(entry.keys())
print("num_points:", n)
print("file_size:", size)
print("features_if_float32:", size // (n * 4))
print("features_if_float64:", size // (n * 8))
```
