```python
def vehicle_ap_aph(metrics):
    # return metrics in the form [l1 ap, l1 aph, l2 ap, l2aph]
    L1ap = metrics['OBJECT_TYPE_TYPE_VEHICLE_LEVEL_1/AP']
    L1aph = metrics['OBJECT_TYPE_TYPE_VEHICLE_LEVEL_1/APH']
    L2ap = metrics['OBJECT_TYPE_TYPE_VEHICLE_LEVEL_2/AP']
    L2aph = metrics['OBJECT_TYPE_TYPE_VEHICLE_LEVEL_2/APH']
    return 100*np.array([L1ap, L1aph, L2ap, L2aph])
def pedestrian_ap_aph(metrics):
    # return metrics in the form [l1 ap, l1 aph, l2 ap, l2aph]
    L1ap = metrics['OBJECT_TYPE_TYPE_PEDESTRIAN_LEVEL_1/AP']
    L1aph = metrics['OBJECT_TYPE_TYPE_PEDESTRIAN_LEVEL_1/APH']
    L2ap = metrics['OBJECT_TYPE_TYPE_PEDESTRIAN_LEVEL_2/AP']
    L2aph = metrics['OBJECT_TYPE_TYPE_PEDESTRIAN_LEVEL_2/APH']
    return 100*np.array([L1ap, L1aph, L2ap, L2aph])
def cyclist_ap_aph(metrics):
    # return metrics in the form [l1 ap, l1 aph, l2 ap, l2aph]
    L1ap = metrics['OBJECT_TYPE_TYPE_CYCLIST_LEVEL_1/AP']
    L1aph = metrics['OBJECT_TYPE_TYPE_CYCLIST_LEVEL_1/APH']
    L2ap = metrics['OBJECT_TYPE_TYPE_CYCLIST_LEVEL_2/AP']
    L2aph = metrics['OBJECT_TYPE_TYPE_CYCLIST_LEVEL_2/APH']
    return 100*np.array([L1ap, L1aph, L2ap, L2aph])
def overall_map_maph(metrics):
    vehicle = vehicle_ap_aph(metrics)
    pedestrian = pedestrian_ap_aph(metrics)
    cyclist = cyclist_ap_aph(metrics)
    l1map = (vehicle[0] + pedestrian[0] + cyclist[0]) / 3
    l1maph = (vehicle[1] + pedestrian[1] + cyclist[1]) / 3
    l2map = (vehicle[2] + pedestrian[2] + cyclist[2]) / 3
    l2maph = (vehicle[3] + pedestrian[3] + cyclist[3]) / 3
    return np.array([round(l1map, 1), round(l1maph, 1), round(l2map, 1), round(l2maph, 1)])
```

```python
def sort_local_metrics_dict(d):
    detector_order = {"msf": 0, "ptt": 1}
    setting_order = {
        "rel_fixed_easy": 0,
        "rel_fixed_medium": 1,
        "rel_fixed_hard": 2,
        "random_easy": 3,
        "random_medium": 4,
        "random_hard": 5,
    }
    timing_order = {
        "clean_hist_pert_obs": 0,
        "pert_hist_pert_obs": 1,
        "pert_hist_clean_obs": 2,
    }

    known_timings = list(timing_order.keys())

    def parse_key(k: str):
        detector, rest = k.split("_", 1)  # ptt, random_medium_pert_hist_pert_obs
        timing = ""
        setting = rest

        for t in known_timings:
            suffix = "_" + t
            if rest.endswith(suffix):
                timing = t
                setting = rest[: -len(suffix)]
                break

        return detector, setting, timing

    def key_fn(item):
        k, _ = item
        detector, setting, timing = parse_key(k)
        return (
            detector_order.get(detector, 999),
            detector,
            setting_order.get(setting, 999),
            setting,
            timing_order.get(timing, 999),
            timing,
            k,
        )

    return dict(sorted(d.items(), key=key_fn))


# usage
# sorted_local_metrics = sort_local_metrics_dict(local_metrics_dict)
sorted_local_metrics = sort_local_metrics_dict(local_metrics_dict)
sorted_local_metrics
```

```python
def split_string(s):
    # 1. Get the prefix (everything before the first _)
    prefix, rest = s.split("_", 1)

    # 2. Find where the "difficulty" ends
    # We look for the keyword and the underscore immediately following it
    for diff in ["_easy_", "_medium_", "_hard_"]:
        if diff in f"_{rest}": # Adding leading _ to ensure we match the full word
            middle, _, suffix = rest.partition(diff.strip("_"))
            # Clean up the middle and suffix
            middle_part = middle + diff.strip("_")
            suffix_part = suffix.lstrip("_")
            break
    return prefix, middle_part, suffix
```

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict

plt.style.use("seaborn-v0_8-paper")

# # ------------------------------------------------
# # organize data by attack mode
# # ------------------------------------------------

attack_groups = defaultdict(dict)

for entry in asr_data:
    
    name = entry["name"]            # e.g. msf_random_easy
    parts = name.split("_")
    
    detector = parts[0]             # msf or ptt
    mode = "_".join(parts[1:])      # random_easy
    
    attack_groups[mode][detector] = np.array(entry["scores_vehicle"])


bins = np.linspace(0,1,21)
difficulties = ["easy","medium","hard"]

for diff in difficulties:

    fig, axes = plt.subplots(
        2, 2,
        figsize=(10,8),
        sharex=True,
        sharey=True
    )

    for r, attack in enumerate(["rel_fixed", "random"]):
        mode = f"{attack}_{diff}"

        if "msf" in attack_groups[mode]:
            axes[r,0].hist(attack_groups[mode]["msf"], bins=bins, edgecolor="black")
            axes[r,0].set_title(f"MSF — {mode}")
            axes[r,0].set_ylabel("Count")

        if "ptt" in attack_groups[mode]:
            axes[r,1].hist(attack_groups[mode]["ptt"], bins=bins, edgecolor="black")
            axes[r,1].set_title(f"PTT — {mode}")

        axes[r,0].axvline(0.5, linestyle="--")
        axes[r,1].axvline(0.5, linestyle="--")

        axes[r,0].set_xlim(0,1)
        axes[r,1].set_xlim(0,1)

    axes[1,0].set_xlabel("Detection confidence")
    axes[1,1].set_xlabel("Detection confidence")

    plt.tight_layout()
    plt.savefig(f"detector_score_{diff}.pdf")
    plt.show()
```

```python
bins = np.linspace(0,1,21)
difficulties = ["easy","medium","hard"]
attacks = ["rel_fixed","random"]

fig, axes = plt.subplots(
    len(difficulties),
    4,
    figsize=(14,10),
    sharex=True,
    sharey=True
)

for i, diff in enumerate(difficulties):

    for j, attack in enumerate(attacks):

        mode = f"{attack}_{diff}"

        # column indices
        col_msf = j*2
        col_ptt = j*2 + 1

        if "msf" in attack_groups[mode]:
            axes[i,col_msf].hist(
                attack_groups[mode]["msf"],
                bins=bins,
                edgecolor="black"
            )
            axes[i,col_msf].set_title(f"MSF — {mode}")

        if "ptt" in attack_groups[mode]:
            axes[i,col_ptt].hist(
                attack_groups[mode]["ptt"],
                bins=bins,
                edgecolor="black"
            )
            axes[i,col_ptt].set_title(f"PTT — {mode}")

        axes[i,col_msf].axvline(0.5, linestyle="--")
        axes[i,col_ptt].axvline(0.5, linestyle="--")

        axes[i,col_msf].set_xlim(0,1)
        axes[i,col_ptt].set_xlim(0,1)

    axes[i,0].set_ylabel(f"{diff}\nCount")

for j in range(4):
    axes[-1,j].set_xlabel("Detection confidence")

plt.tight_layout()
plt.savefig("detector_score_all.pdf")
plt.show()
```

```python
import pandas as pd

difficulties = ["easy", "medium", "hard"]
attacks = ["rel_fixed", "random"]

rows = []

for diff in difficulties:
    row = {"difficulty": diff}

    for attack in attacks:
        mode = f"{attack}_{diff}"

        for det in ["msf", "ptt"]:
            scores = attack_groups[mode][det]

            pct = np.mean(scores >= 0.5) * 100
            row[f"{attack}_{det.upper()}"] = pct

    rows.append(row)

df = pd.DataFrame(rows)
df = df.set_index("difficulty")

print(df.round(1))
```

```python
import pandas as pd

# -----------------------------
# helper formatting
# -----------------------------
def format_percent(x):
    x = x * 100
    if x < 1:
        return f"{x:.1g}%"
    else:
        return f"{x:.1f}%"


# -----------------------------
# organize data
# -----------------------------
rows = {}

for entry in asr_data:
    
    name = entry["name"]
    parts = name.split("_")
    
    detector = parts[0]                  # msf / ptt
    mode = "_".join(parts[1:])           # random_easy etc
    
    if mode not in rows:
        rows[mode] = {}
    
    rows[mode][(detector,"vehicle")] = entry["asr_vehicle"]
    rows[mode][(detector,"pedestrian")] = entry["asr_pedestrian"]
    rows[mode][(detector,"cyclist")] = entry["asr_cyclist"]


# -----------------------------
# desired row order
# -----------------------------
row_order = [
    "random_easy",
    "rel_fixed_easy",
    "random_medium",
    "rel_fixed_medium",
    "random_hard",
    "rel_fixed_hard"
]


# -----------------------------
# build dataframe
# -----------------------------
data = []

for mode in row_order:
    
    row = rows.get(mode,{})
    
    data.append([
        format_percent(row.get(("msf","vehicle"),0)),
        format_percent(row.get(("ptt","vehicle"),0)),
        format_percent(row.get(("msf","pedestrian"),0)),
        format_percent(row.get(("ptt","pedestrian"),0)),
        format_percent(row.get(("msf","cyclist"),0)),
        format_percent(row.get(("ptt","cyclist"),0)),
    ])


columns = pd.MultiIndex.from_tuples([
    ("MSF","Vehicle"),
    ("PTT","Vehicle"),
    ("MSF","Pedestrian"),
    ("PTT","Pedestrian"),
    ("MSF","Cyclist"),
    ("PTT","Cyclist"),
])


df = pd.DataFrame(data, index=row_order, columns=columns)

print(df)
```
