```python
def points_in_box_3d(points_xyz: np.ndarray, box: np.ndarray) -> np.ndarray:
    """
    Return boolean mask of which points fall inside an oriented 3D box.

    Parameters
    ----------
    points_xyz : (N, 3)  x, y, z
    box        : (7,)    cx, cy, cz, dx, dy, dz, heading
                         dx/dy/dz are full extents (not half-extents)
                         heading is yaw in radians

    Returns
    -------
    mask : (N,) bool
    """
    cx, cy, cz, dx, dy, dz, heading = box[:7]

    # translate points to box centre
    pts = points_xyz.copy()
    pts[:, 0] -= cx
    pts[:, 1] -= cy
    pts[:, 2] -= cz

    # rotate into box-aligned frame (undo heading)
    cos_h = np.cos(-heading)
    sin_h = np.sin(-heading)
    x_rot =  cos_h * pts[:, 0] - sin_h * pts[:, 1]
    y_rot =  sin_h * pts[:, 0] + cos_h * pts[:, 1]
    z_rot =  pts[:, 2]

    # check half-extents
    mask = (
        (np.abs(x_rot) <= dx / 2.0) &
        (np.abs(y_rot) <= dy / 2.0) &
        (np.abs(z_rot) <= dz / 2.0)
    )
    return mask

def count_points_in_box(points: np.ndarray, box: np.ndarray) -> int:
    """points is (N, >=3); uses first 3 cols as xyz."""
    if points.shape[0] == 0:
        return 0
    mask = points_in_box_3d(points[:, :3], box)
    return int(mask.sum())
```

```python

global_to_segment = {}
segment_frame_counters = {}
for i, info in enumerate(dataset.infos):
    seg = info['point_cloud']['lidar_sequence']
    if seg not in segment_frame_counters:
        segment_frame_counters[seg] = 0
    global_to_segment[i] = (seg, segment_frame_counters[seg])
    segment_frame_counters[seg] += 1
    


dataset_files = os.listdir('datasets/rel_fixed_hard_d')

ordered_segments = [info['point_cloud']['lidar_sequence'] for info in dataset.infos]
ordered_segments = list(dict.fromkeys(ordered_segments))  # remove duplicates while preserving order

# map filenames to segment names
file_map = {f.replace('_d.pkl',''): f for f in dataset_files}
# reorder files
dataset_files_ordered = [file_map[s] for s in ordered_segments if s in file_map]

seg_dataset_file = f"datasets/rel_fixed_hard_d/{dataset_files_ordered[0]}"
# Open the file in read-binary mode ('rb')
with open(seg_dataset_file, 'rb') as file:
    # Use pickle.load() to deserialize the data
    seg_dataset = pkl.load(file)
```

```python
current_segment = None
for j in range(38597):
    
    segment = dataset[j]['frame_id'].rsplit("_", 1)[0]
    seg_idx = global_to_segment[j][1]
    if current_segment != segment:

        # save previous segment if it exists

        current_segment = segment
        print(current_segment)
        seg_dataset_file = f"datasets/rel_fixed_hard_d/{current_segment}_d.pkl"
        with open(seg_dataset_file, "rb") as f:
            seg_dataset = pkl.load(f)
    assert dataset[j]['frame_id'] == seg_dataset[seg_idx]['frame_id']

    gt = dataset[j]['gt_boxes']
    points_clean = dataset[j]['points'][:, :3]
    points_spoof = seg_dataset[seg_idx]['points'][:, :3]
    cyc_indices_gt = [i for i, val in enumerate(gt[:, -1]) if val == 3]
    counts_clean = []
    counts_spoof = []
    if cyc_indices_gt:
        for cyc_idx in cyc_indices_gt:
            cyc_box = gt[cyc_idx]
            count_clean = count_points_in_box(points_clean, cyc_box)
            count_spoof = count_points_in_box(points_spoof, cyc_box)
            counts_clean.append(count_clean)
            counts_spoof.append(count_spoof)
        diff = [c - s for (c, s) in zip(counts_clean, counts_spoof)]
        print(j, diff)
   
```
