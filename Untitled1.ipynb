```python
import numpy as np
import tensorflow as tf
from waymo_open_dataset import dataset_pb2 as open_dataset
import zlib

TFRECORD_PATH = 'individual_files_validation_segment-10203656353524179475_7625_000_7645_000_with_camera_labels.tfrecord'
MAX_FRAMES = None  # limit for testing (set None for full file)
OUTPUT_PREFIX = "waymo_stats"

RMIN = 6
RMAX = 12

def decompress(ri_compressed):
    ri = open_dataset.MatrixFloat()
    ri.ParseFromString(zlib.decompress(ri_compressed))
    return np.array(ri.data).reshape(ri.shape.dims)

intensity_vals = []
elongation_vals = []

total_band_rays = 0
total_two_return = 0

dataset = tf.data.TFRecordDataset(TFRECORD_PATH, compression_type='')

for frame_idx, data in enumerate(dataset):

    if MAX_FRAMES is not None and frame_idx >= MAX_FRAMES:
        break

    frame = open_dataset.Frame()
    frame.ParseFromString(bytearray(data.numpy()))

    laser = frame.lasers[open_dataset.LaserName.TOP]

    ri1 = decompress(laser.ri_return1.range_image_compressed)
    ri2 = decompress(laser.ri_return2.range_image_compressed)

    r1 = ri1[..., 0]
    I1 = ri1[..., 1]
    E1 = ri1[..., 2]

    r2 = ri2[..., 0]

    mask1 = r1 > 0
    mask2 = r2 > 0

    # Rays whose FIRST return is in band
    band_mask = mask1 & (r1 >= RMIN) & (r1 <= RMAX)

    intensity_vals.append(I1[band_mask])
    elongation_vals.append(E1[band_mask])

    num_band = np.sum(band_mask)
    num_two = np.sum(band_mask & mask2)

    total_band_rays += num_band
    total_two_return += num_two

    # print(f"Processed frame {frame_idx}")
    
intensity_vals = np.concatenate(intensity_vals)
elongation_vals = np.concatenate(elongation_vals)

mean_intensity = intensity_vals.mean()
std_intensity = intensity_vals.std()

mean_elongation = elongation_vals.mean()
std_elongation = elongation_vals.std()

P_two_return = total_two_return / total_band_rays

print("\n===== Results =====")
print(f"Range band: [{RMIN}, {RMAX}] m")
print(f"Samples: {len(intensity_vals)}")
print(f"Intensity mean: {mean_intensity}")
print(f"Intensity std:  {std_intensity}")
print(f"Elongation mean: {mean_elongation}")
print(f"Elongation std:  {std_elongation}")
print(f"P(2 returns | first return in band): {P_two_return}")
```

```python
import numpy as np
import tensorflow as tf
from waymo_open_dataset import dataset_pb2 as open_dataset
import zlib

TFRECORD_PATH = 'individual_files_validation_segment-10203656353524179475_7625_000_7645_000_with_camera_labels.tfrecord'
MAX_FRAMES = None

# ----------------------------
# Define range bins
# ----------------------------
range_bins = [
    (6, 12),
    (12, 18),
    (18, 24),
    (24, 30),
    (30, 36),
    (36, 42)
]

# Initialize storage
band_data = {
    band: {
        "intensity": [],
        "elongation": [],
        "total_band_rays": 0,
        "total_two_return": 0
    }
    for band in range_bins
}

def decompress(ri_compressed):
    ri = open_dataset.MatrixFloat()
    ri.ParseFromString(zlib.decompress(ri_compressed))
    return np.array(ri.data).reshape(ri.shape.dims)

dataset = tf.data.TFRecordDataset(TFRECORD_PATH, compression_type='')

for frame_idx, data in enumerate(dataset):

    if MAX_FRAMES is not None and frame_idx >= MAX_FRAMES:
        break

    frame = open_dataset.Frame()
    frame.ParseFromString(bytearray(data.numpy()))

    laser = frame.lasers[open_dataset.LaserName.TOP]

    ri1 = decompress(laser.ri_return1.range_image_compressed)
    ri2 = decompress(laser.ri_return2.range_image_compressed)

    r1 = ri1[..., 0]
    I1 = ri1[..., 1]
    E1 = ri1[..., 2]
    r2 = ri2[..., 0]

    mask1 = r1 > 0
    mask2 = r2 > 0

    for (RMIN, RMAX) in range_bins:

        band_mask = mask1 & (r1 >= RMIN) & (r1 <= RMAX)

        band_data[(RMIN, RMAX)]["intensity"].append(I1[band_mask])
        band_data[(RMIN, RMAX)]["elongation"].append(E1[band_mask])

        num_band = np.sum(band_mask)
        num_two = np.sum(band_mask & mask2)

        band_data[(RMIN, RMAX)]["total_band_rays"] += num_band
        band_data[(RMIN, RMAX)]["total_two_return"] += num_two

# ----------------------------
# Compute final stats
# ----------------------------
final_stats = {}

for band in range_bins:

    intensity_vals = np.concatenate(band_data[band]["intensity"])
    elongation_vals = np.concatenate(band_data[band]["elongation"])
    if len(intensity_vals) < 10:
        print(f"Skipping band {band} (insufficient samples)")
        continue
    # ----- Intensity (log-normal model) -----
    positive_intensity = intensity_vals[intensity_vals > 0]
    log_intensity = np.log(positive_intensity)

    log_mu = log_intensity.mean()
    log_sigma = log_intensity.std(ddof=1)

    # Robust percentiles (optional safety bounds)
    p1, p99 = np.percentile(positive_intensity, [1, 99])

    # ----- Elongation (empirical pool) -----
    elong_pool = elongation_vals.astype(np.float32)

    # ----- Double return probability -----
    total_band = band_data[band]["total_band_rays"]
    total_two = band_data[band]["total_two_return"]
    P_two = total_two / total_band if total_band > 0 else 0.0

    final_stats[band] = {
        "log_mu": float(log_mu),
        "log_sigma": float(log_sigma),
        "intensity_p1": float(p1),
        "intensity_p99": float(p99),
        "elong_pool": elong_pool,
        "P_two": float(P_two),
        "n_samples": int(len(intensity_vals))
    }

    print("\n===== Band:", band, "=====")
    print("Samples:", len(intensity_vals))
    print("log_mu:", log_mu)
    print("log_sigma:", log_sigma)
    print("P_two:", P_two)

np.save("range_conditioned_stats.npy", final_stats, allow_pickle=True)

print("\nSaved range-conditioned statistics.")
```
