```python
root = "/storage/cedar/cedar0/cedarp-gchou3-0/data/waymo"
tag = "waymo_processed_data_v4"

with open(os.path.join(root, "waymo_infos_train_v4.pkl"), "rb") as f:
    global_infos = pickle.load(f)

ordered_infos = []
with open(os.path.join(root, "ImageSets", "train.txt")) as f:
    for line in f:
        seq = os.path.splitext(line.strip())[0]
        pkl_path = os.path.join(root, tag, seq, f"{seq}.pkl")
        with open(pkl_path, "rb") as pf:
            ordered_infos.extend(pickle.load(pf))

print("global len:", len(global_infos))
print("ordered len:", len(ordered_infos))

for k in [0, 1, 10, 100, 1000, 10000, 50000, 100000, len(global_infos)-1]:
    a = global_infos[k]
    b = ordered_infos[k]
    same = (a["sequence_name"] == b["sequence_name"] and a["sample_idx"] == b["sample_idx"])
    print(k, same, (a["sequence_name"], a["sample_idx"]), (b["sequence_name"], b["sample_idx"]))
```

```python
import os, pickle

root = "/storage/cedar/cedar0/cedarp-gchou3-0/data/waymo"
tag = "waymo_processed_data_v4"

infos = []
with open(os.path.join(root, "ImageSets", "train.txt")) as f:
    for line in f:
        seq = os.path.splitext(line.strip())[0]
        pkl_path = os.path.join(root, tag, seq, f"{seq}.pkl")
        with open(pkl_path, "rb") as pf:
            infos.extend(pickle.load(pf))

with open(os.path.join(root, "waymo_infos_train_v4.pkl"), "wb") as f:
    pickle.dump(infos, f)

print("Rebuilt ordered infos:", len(infos))
```
